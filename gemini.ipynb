```python
import os
from vanna.flask import VannaFlaskApp
import vanna as vn

# Customizing the VannaFlaskApp to only generate SQL without execution
class CustomVannaFlaskApp(VannaFlaskApp):
    def generate_sql(self, question: str):
        sql_query = vn.generate_sql(question=question)
        return {"sql": sql_query}

# Create the Flask app using the custom class
app = CustomVannaFlaskApp(vn)

# Run the app
if __name__ == "__main__":
    # Run the app and print the assigned port
    from werkzeug.serving import run_simple
    server = run_simple("127.0.0.1", 0, app, use_reloader=False, use_debugger=True)
    print(f"App is running on port: {server.port}")
    app.run(debug=True, host="127.0.0.1", port=0, use_reloader=False)
```
